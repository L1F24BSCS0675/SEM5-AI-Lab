```python
# TASK1
# CONVERT INTEGERS TO ITS ROMAN NUMERAL
# first of all we have to make its class
# we have to initialize the integer using constructor 
# class

class rNumeral:
    def __init__(self, number): #INITIALIZE
        self.number = number

    def to_roman(self):    # function to convert in roman
        values = [ 
            (1000, "M"),
            (900, "CM"),
            (500, "D"),
            (400, "CD"),
            (100, "C"),
            (90, "XC"),
            (50, "L"),
            (40, "XL"),
            (10, "X"),
            (9, "IX"),
            (5, "V"),
            (4, "IV"),
            (1, "I")
        ]
        num = self.number
        result = ""

        for value, symbol in values:
            while num >= value:
                result += symbol
                num -= value

        return result
        
obj = rNumeral(3549) # this is a object
print(obj.to_roman()) # this is the function call to convert digit to roman numeral

```

```python
# TASK2
# TO CHECK THAT TWO STRINGS ARE ANAGRAMS OF EACH OTHER OR NOT 

def check_anagrams(text1, text2):
    return sorted(text1.lower()) == sorted(text2.lower()) 

print(check_anagrams("listen", "silent"))  
```
