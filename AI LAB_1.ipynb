```python
# Q2
class Roman:
    def convert(self,num):
        values = [1000, 900, 500, 400, 100, 90, 50, 40, 10, 9, 5, 4, 1]
        symbols = ["M", "CM", "D", "CD", "C", "XC", "L", "XL",
                   "X", "IX", "V", "IV", "I"]

        roman = ""

        for i in range(len(values)):
            while num >= values[i]:
                roman = roman + symbols[i]
                num = num - values[i]

        return roman


r = Roman()
print(r.convert(25))
                 
    
    
```

```python
# Q3
def pair(digits,target):
    for i in range(len(digits)):
        for j in range(i+1,len(digits)):
            if digits[i]+digits[j]==target:
                print("(", i, ",", j,")")

digits=[10,20,30,11,60]
target=21
pair(digits,target)
```

```python
#  Q4
class eor():
    def fnd(self, digits):
        R = []

        for i in range(len(digits)):
            for j in range(i + 1, len(digits)):
                for k in range(j + 1, len(digits)):
                    if digits[i] + digits[j] + digits[k] == 0:
                        R.append([digits[i], digits[j], digits[k]])

        return R


e = eor()

digits = [10, -10, 0, 3, 2,-3]

print(e.fnd(digits))

```

```python
#  Q10
class Numbers():
    multiplyer=2
    def __init__(self,x,y):
        self.x=x
        self.y=y
        
    def add(self):
        return self.x+self.y

    @classmethod    # use the class atribute
    def mul(cls,a):
        return a*cls.multiplyer

    @staticmethod  
    def sub(b,c):
        return b-c

    @property  # no we can access the vlue as an atribute we dont need to use ()
    def value(self):
        return (self.x,self.y)   # return a tuple (,)

    @value.setter 
    def value(self,new):
        self.x=new[0]
        self.y=new[1]

    
    @value.deleter
    def value(self):
        del self.x
        del self.y
        


n=Numbers(1,2)
print(n.add())

print(n.mul(2))

print(n.sub(3,2))

print(n.value)

# setter by using property it will converted into property thats why it  is directly setable 
n.value = (50, 60)
print(n.value)

# deleter
del n.value

    
    
    
        
```

```python
# Q8
matrix1 = [
    [1, 2],
    [3, 4]
]

matrix2 = [
    [5, 6],
    [7, 8]
]

result = [
    [0, 0],
    [0, 0]
]

for i in range(2):
    for j in range(2):
        for k in range(2):
            result[i][j] = result[i][j] + matrix1[i][k] * matrix2[k][j]

print("Result of multiplication:")

for row in result:
    print(row)
```

```python
# Q9
def calculator(num1, num2, operation="addition", output="floating"):

    if operation == "addition":
        result = num1 + num2

    elif operation == "subtraction":
        result = num1 - num2

    elif operation == "multiplication":
        result = num1 * num2

    elif operation == "division":
        if num2 == 0:
            raise Exception("Cannot divide by zero")
        result = num1 / num2

    else:
        raise Exception("Invalid operation")

    if output == "integer":
        return round(result)

    elif output == "floating":
        return float(result)

    else:
        raise Exception("Invalid output format")


print(calculator(10, 5, "addition", "integer"))
print(calculator(10, 3, "division", "floating"))
print(calculator(10, 3, "division", "integer"))
```

```python
# Exercise 1.1 
def count_pattern(pattern, lst):
    count = 0

    for i in range(len(lst) - 1):
        if lst[i] == pattern[0] and lst[i + 1] == pattern[1]:
            count = count + 1

    return count


print(count_pattern(('a', 'b'), ('a', 'b', 'c', 'e', 'b', 'a', 'b', 'f')))
```
