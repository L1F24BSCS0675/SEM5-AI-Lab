```python
# greedy best first search 
graph={
'A':{'B':1,'C':3},
'B':{'D':4,'E':5},
'C':{'F':3,'G':6},
'D':{},
'E':{},
'F':{},
'G':{}
}
heuristic={
 'A':8,
  'B':6,
'C':5,
'D':4,
'E':4,
'F':3,
'G':0

}
import heapq
def gbfs(graph,start,goal):
    cost=0
    path=[]
    visited=set()
    heap=[(heuristic[start],start,path)]
    while heap:
        cost,current,path=heapq.heappop(heap)
        if current==goal:
            path.append(current)
            return path,cost
        if current not in visited:
            visited.add(current)
            path=path+[current]
            for x,y in graph.get(current,{}).items():
                heapq.heappush(heap,(heuristic[x],x,path))
    return None,float('info')

gbfs(graph,'A','G')
    
```

```python
# A* search
graph={
'A':{'B':1,'C':3},
'B':{'D':4,'E':5},
'C':{'F':3,'G':6},
'D':{},
'E':{},
'F':{},
'G':{}
}


heuristi={
 'A':8,
  'B':6,
'C':5,
'D':4,
'E':4,
'F':3,
'G':0

}
import heapq
def astar(start,goal):
    cost=0
    path=[]
    visited=set()
    heap=[(heuristic[start],cost,start,path)]
    while heap:
        st_cost,ac_cost,current,path=heapq.heappop(heap)
        if current==goal:
            path.append(current)
            return path,ac_cost
        if current not in visited:
            visited.add(current)
            path=path+[current]
            for x,y in graph.get(current,{}).items():
                new_ac_cost=ac_cost+y
                new_st_cost=new_ac_cost+heuristic[x]
                heapq.heappush(heap,(new_st_cost,new_ac_cost,x,path))
    return None,float('inf')

astar('A','G')









```
