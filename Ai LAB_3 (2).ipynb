```python
import numpy as np
normal_list=[1,2,3,4,5]
nump_list=np.array([1,2,3,4,5])
print("Normal python list ",normal_list)
print("Numpy list ",nump_list)
print("Numpy list with + 2  ",nump_list+2)# add 2 in every digit in numpy 
print("Normal list with + 2  ",normal_list+[2])# it will dont add 2 but append 2 at the last of the array ni the normal python list 
arr0 =np.zeros(5)# array created and initalze with 0
arr1 =np.ones(5)# array created and initalze with 1
arr2=np.arange(1,11)# starting and end range with .arrange
arr3=np.arange(0,20,2)# increment of 2 
print("Array0=",arr0)
print("Array1=",arr1)
print("Array2=",arr2)
print("Array3=",arr3)
# print by index array 
marks=np.array([75,82,11,100])
marks1=np.array([5,82,1,10])
print("total marks+marks1 after add ",marks+marks1)
print("total marks-marks1 after sub ",marks-marks1)

print("First index element= ",marks[1])
print("Last  index element= ",marks[-1])
print(marks[1:2])#  1 index
print(marks[:2])# 0 to 1 ---index  before 2 all elements 
print(marks[1:])#  also include last --- after 1 index all
res=np.array([1,2,3,4,5])
print("Sum= ",np.sum(res))
print("Mean= ",np.mean(res))
print("Median= ",np.median(res))
print("Min= ",np.min(res))
print("Max= ",np.max(res))
print("Standard Deviation= ",np.std(res))


```

```python
data={ "Name":["Ali","sara","Adi"],"AI":[12,56,1],"Maths":[1,2,45]}
# Dataframe means table 

print("Print the data in tables form",pd.DataFrame(data))
f=pd.DataFrame(data)
print("Shape of the array=",f.shape) # rows and columns 
print("column of the array=",f.columns)
print("datatype of the array=",f.dtypes)
print("Information ",f.info())
```

```python
#
data={ "Name":["Ali","sara","Adi"],"marks":[12,56,1],"age":[1,2,45]}
df=pd.DataFrame(data,index=["a","b","c"]) # CHANGE THE INDEX 
print(df)
# loc function is used when we rechange the index and then access data by using the user defined index 
print(df.loc["c","Name"]) # index are a b c for every 
print(df.loc["a","marks"])
print("Sum of the marks",df["marks"].sum()) # sum of seprate rows 
```
