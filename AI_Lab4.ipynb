```python
# bfs 
graph={
    'A':{'B':1,'C':2},
    'B':{'D':3,'E':6},
    'C':{'F':4,'G':5},
    'D':{},
    'E':{},
    'F':{},
    'G':{},
}
def bfs(graph,start,goal):
    cost=0# because 1 node have a 0 cost so set it manually 
    path=[]
    visited=set() # set to keep the explored unique 
    queue=[(start,cost)]   # queue fun which store the tuple here first value is enque in the queue 
    while queue:# until queue is empty 
        current,cost=queue.pop(0) #  pop from 1 index value with the cost 
        if(current==goal):# if the goal found 
            path.append(current)
            return path,cost # when goal found return the fun 
        if current not in visited:
            path.append(current)
            visited.add(current)#  nodes who are explored 
            for x,y in graph.get(current,{}).items():# second {} used coz we use here dict inside dict # x neighbours y cost # used to get neighbours of the current but if there is no any neighbour of the node  for loop not  run 
                queue.append((x,y+cost))# double brackets 1 for tuple and 1 for fun y+cost means calculate the sum of all prev nodes to reach till the curr node  
    return None,float('inf') #  to handle the error if the graph is not correct 

bfs(graph,'A','G')



                
            
            
        
        
```

```python
# bfs 
graph={
    'A':{'B':1,'C':2},
    'B':{'D':3,'E':6},
    'C':{'F':4,'G':5},
    'D':{},
    'E':{},
    'F':{},
    'G':{},
}
def dfs(graph,start,goal):
    cost=0# because 1 node have a 0 cost so set it manually 
    path=[]
    visited=set() # set to keep the explored unique 
    stack=[(start,cost)]   # queue fun which store the tuple here first value is enque in the queue 
    while stack:# until queue is empty 
        current,cost=stack.pop() #  pop from 1 index value with the cost 
        if(current==goal):# if the goal found 
            path.append(current)
            return path,cost # when goal found return the fun 
        if current not in visited:
            path.append(current)
            visited.add(current)#  nodes who are explored 
            for x,y in graph.get(current,{}).items():# second {} used coz we use here dict inside dict # x neighbours y cost # used to get neighbours of the current but if there is no any neighbour of the node  for loop not  run 
                stack.append((x,y+cost))# double brackets 1 for tuple and 1 for fun y+cost means calculate the sum of all prev nodes to reach till the curr node  
    return None,float('inf') #  to handle the error if the graph is not correct 

dfs(graph,'A','G')




                
            
     
```

```python
graph={
    'A':{'B':1,'C':2},
    'B':{'D':3},
    'C':{'D':1},
    'D':{},   
}

# USE MINHAP TO PICK FIRST LOWERCOST 
import heapq
def ucs(graph,start,goal):
    cost=0
    path=[]
    visited=set()
    heap=[(cost,start,path)]# minheap check 0 index on the base of cost it will go to the shorter path 
    while heap:
        cost,current,path=heapq.heappop(heap)# use heappop fun by using heapq library 
        if current==goal:
            path.append(current)
            return path,cost
        if current not in visited:
            visited.add(current)
            path=path+[current]# add path with the previous coz may be path change 
            for x,y in graph.get(current,{}).items():
                heapq.heappush(heap,(y+cost,x,path))
    return None,float('info')

ucs(graph,'A','D') 
```
